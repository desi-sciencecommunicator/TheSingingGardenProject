```python
import os 
import matplotlib.pyplot as plt 
import numpy as np
import wave
```

```python
# shows the sound waves
def visualize(path: str):
  
    # reading the audio file
    raw = wave.open(path)
    
    # reads all the frames 
    # -1 indicates all or max frames
    signal = raw.readframes(-1)
    signal = np.frombuffer(signal, dtype ="int16")
    
    # gets the frame rate
    f_rate = raw.getframerate()

    # to Plot the x-axis in seconds 
    # you need get the frame rate 
    # and divide by size of your signal
    # to create a Time Vector 
    # spaced linearly with the size 
    # of the audio file
    time = np.linspace(
        0, # start
        len(signal) / f_rate,
        num = len(signal)
    )

    # using matplotlib to plot
    # creates a new figure
    plt.figure(figsize=(3.5, 2))
    
    # title of the plot
    plt.title("Sound Wave")
    
    # label of x-axis
    plt.xlabel("Time")
   
    # actual plotting
    plt.plot(time, signal, linewidth=.15, color='black')
    plt.savefig('/MYPATH/wave.svg')
    # shows the plot 
    # in new window
    plt.show()

    # you can also save
    # the plot using
    # plt.savefig('filename')
```

```python
# USER INPUTTED PARAMATERS

data_dir = '/MYPATH/Recordings/20260817_to20260818/' #include date if looking for that 
data_fname = os.path.join(data_dir, 'TRIMMED_EXAMPLE.wav')
# print('DATA FILE {0}\nEXISTS: {1}'.format(data_fname, os.path.isfile(data_fname)))
```

```python
visualize(data_fname)
```
