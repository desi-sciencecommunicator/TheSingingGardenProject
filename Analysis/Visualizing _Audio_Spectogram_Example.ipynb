```python
import os
# for data transformation 
import numpy as np 
# for visualizing the data 
import matplotlib.pyplot as plt 
# for opening the media file 
import scipy.io.wavfile as wavfile
```

```python
# USER INPUTTED PARAMATERS

data_dir = '/MYPATH/Recordings/20260817_to20260818/' #include date if looking for that 
data_fname = os.path.join(data_dir, 'TRIMMED_EXAMPLE.wav')
# print('DATA FILE {0}\nEXISTS: {1}'.format(data_fname, os.path.isfile(data_fname)))
```

```python
Fs, aud = wavfile.read(data_fname) 
first = aud[:int(Fs*400)] # trim the first 125 seconds
```

```python
plt.figure(figsize=(3.5, 1))
powerSpectrum, frequenciesFound, time, imageAxis = plt.specgram(first, cmap='magma', Fs=Fs) 
plt.xlabel('time')
plt.axis('off')
plt.title('Spectogram of audio file')
plt.savefig('/MYPATH/spectrogram.svg', dpi=300)
plt.show()
```
